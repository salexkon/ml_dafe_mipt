# baseline_naive

Этот пример не обучает сеть. Он проверяет путь от `sample_submission.csv` до
корректного файла отправки: каждой картинке назначается класс `bart_simpson`.
Бейзлайн для домашней работы нужно превзойти обученной моделью.

В Kaggle Notebook сначала подключите данные соревнования через панель Input.
Для локального запуска можно задать путь к образцу переменной окружения
`HW3_SAMPLE_SUBMISSION`.

```python
import os
from pathlib import Path
import pandas as pd
```

```python
configured_path = os.environ.get('HW3_SAMPLE_SUBMISSION')
if configured_path:
    sample_path = Path(configured_path)
else:
    candidates = sorted(Path('/kaggle/input').rglob('sample_submission.csv'))
    if len(candidates) != 1:
        raise ValueError('Подключите данные одного соревнования или задайте HW3_SAMPLE_SUBMISSION.')
    sample_path = candidates[0]
sample_path
```

```python
sample = pd.read_csv(sample_path, dtype={'Id': str})
sample.head()
```

Сохраняем заголовки и идентификаторы из образца. Меняется только столбец
предсказаний. Имя класса пишется полностью: `bart_simpson`.

```python
assert list(sample.columns) == ['Id', 'Expected']
assert sample['Id'].notna().all() and sample['Id'].is_unique
submission = sample.copy()
submission['Expected'] = 'bart_simpson'
submission.head()
```

```python
submission_path = Path('baseline_naive.csv')
submission.to_csv(submission_path, index=False)
print(f'{submission_path.resolve()}: {len(submission)} строк')
```

```python
check = pd.read_csv(submission_path, dtype={'Id': str})
assert list(check.columns) == ['Id', 'Expected']
assert check['Id'].equals(sample['Id'])
assert check['Expected'].eq('bart_simpson').all()
check.head()
```

Полученный `baseline_naive.csv` отправляют в соревнование через
**Submit Predictions**. Оценка появляется только после проверки файла Kaggle.
Для своей модели замените константу предсказанными именами классов,
сохранив соответствие между `Id` и картинкой.
