# HW3. Journey to Springfield: классификация персонажей

Постройте модель, которая определяет персонажа «Симпсонов» по изображению.
Архитектуру и подход выбираете сами. На очной защите покажите свой код,
результат на лидерборде и эксперименты, которыми обосновали выбор модели.

## Данные и соревнование

Данные взяты из учебного соревнования Journey to Springfield школы DLS.
[Соревнование курса](https://www.kaggle.com/competitions/mipt-dafe-ml-2026-simpsons),
[ссылка для вступления](https://www.kaggle.com/t/a57d062186f245dab0a4c89f28cf4250).
Для курса используется отдельный лидерборд.
Набор изображений, задача и метрика сохраняются как в исходном соревновании.

В `train/train/<имя_класса>/` находятся 20 933 обучающих изображения 42 персонажей.
В `test/testset/` находится 991 тестовое изображение. Для локальной работы
скачайте общий архив на вкладке **Data** и распакуйте его.
Обучающие изображения сгруппированы по папкам: имя папки - имя класса.
Тестовые изображения идут без меток. Файл `sample_submission.csv` задаёт
идентификаторы тестовых объектов и формат ответа. Сохраняйте их без изменений.

В настройках исходного соревнования метрика называется **F-Score Beta (Micro)**.
При одном правильном и одном предсказанном классе на объект micro-F1 совпадает
с accuracy: верные ответы дают TP, а каждый неверный одновременно даёт FP и FN.
Macro-F1 усредняет F1 отдельно по классам и может заметно отличаться.
Для выбора модели ориентируйтесь на ту же метрику, что считается в соревновании.

Публичный рейтинг использует 50 изображений, приватный - 941.
Идентификаторы обеих частей фиксированы между посылками. Доступно две посылки
в сутки; для итогового приватного рейтинга можно выбрать две посылки.
Закрытые ответы не входят в скачиваемые файлы.

[Инструкция по отправке со скриншотами](submission_guide.md).

### Бейзлайны в публичном рейтинге

| Бейзлайн | Публичный micro-F1 |
|:--|--:|
| `baseline_naive` | 0.04000 |
| `baseline_mlp` | 0.30000 |
| `baseline_cnn` | 0.58000 |

[Проверка формата наивным бейзлайном](https://www.kaggle.com/code/ampiro/mipt-dafe-ml-2026-baseline-naive).


## Что сделать

**Модель.** Реализуйте свой способ классификации изображений. Сохраните
конфигурацию и обученную модель, чтобы повторить оценку и получить тот же CSV.

**Валидация.** Отделите часть размеченных данных и используйте одно разбиение
для сравниваемых вариантов. Опишите, как получили разбиение и почему оно
подходит для задачи. Используйте локальный micro-F1 для выбора модели.

**Эксперименты.** Сформулируйте, что проверяете, и сравните несколько вариантов.
Для каждого зафиксируйте архитектуру или тип модели, гиперпараметры, способ
обучения, локальную метрику и время обучения. Если используется оптимизатор,
укажите его настройки. Покажите графики, по которым можно сравнить запуски,
матрицу ошибок и примеры ошибок итоговой модели. Объясните выбор модели
по результатам этих экспериментов.

**Отправка.** Сформируйте CSV для всех тестовых изображений и сделайте
успешную посылку на Kaggle. Свяжите её с сохранённой моделью, конфигурацией
и версией кода. Идентификатор каждой строки должен соответствовать изображению.

## Что сдавать

Оформите папку `hw3/` в своём репозитории и откройте к ней доступ преподавателю.

1. Сделайте хотя бы одну успешную отправку на Kaggle. В отчёте укажите команду, идентификатор посылки или ссылку на неё и публичный скор.
2. Одна посылка итоговой модели должна **строго превысить все три бейзлайна** по публичному micro-F1. Текущий порог — **выше 0.58000**.
3. Сдайте оформленный код обучения, оценки на валидации и инференса итоговой модели. Приложите конфигурацию, сохранённую модель и команды запуска.
4. Сдайте отчёт с таблицей, графиками и анализом экспериментов. Для каждого запуска укажите архитектуру, гиперпараметры, способ обучения и результаты. Объясните, какие изменения помогли и как выбрали итоговую модель.
5. Оформите проект по [шаблону HW3](project_template.md).

## Защита

Покажите подготовку данных, обучение, локальную оценку и инференс итоговой
модели. Объясните, почему выбрали этот подход и что показали эксперименты.
Будьте готовы загрузить сохранённую модель и воспроизвести посылку.

## Оценка

**Условие приёма.** Хотя бы одна успешная посылка итоговой модели должна строго превысить все три бейзлайна по публичному micro-F1. При результате 0.58 или ниже работа не принимается.

| Публичный micro-F1 итоговой посылки | Баллы за скор |
|:--|--:|
| Выше 0.95 | 5 |
| От 0.90 до 0.95 включительно | 4 |
| От 0.85 до 0.90, верхняя граница исключена | 3 |
| От 0.80 до 0.85, верхняя граница исключена | 2 |
| Выше 0.58 и ниже 0.80 | 1 |
| 0.58 и ниже | Работа не принимается |

За принятую работу, помимо баллов за скор, начисляется:

- **До 2 баллов за валидацию.** Обоснуйте фиксированное разбиение, считайте micro-F1 на отложенных данных и используйте валидацию для выбора модели.
- **До 3 баллов за собственные эксперименты.** Сравните варианты в одинаковых условиях, покажите таблицу и графики. Обоснуйте выбор итоговой модели результатами. Укажите, что меняли в архитектуре, гиперпараметрах и способе обучения, включая оптимизатор, если он используется.

Максимум за HW3 — **10 баллов**.

Ответы теста нельзя подменять найденной разметкой или вручную исправлять
по отдельным изображениям. Работать можно командой до четырёх человек.
На защите каждый участник должен понимать все части решения.
